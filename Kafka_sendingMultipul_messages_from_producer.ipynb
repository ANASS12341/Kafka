```python

from confluent_kafka import Producer
import json
import time
# Initialisation du Producteur Kafka
producer = Producer(kafka_config)
```

```python
import json

topic = "ecommerce"
with open('customers.json', 'r') as file:
    # Charger la chaîne JSON doublement sérialisée
    raw_data = json.load(file)
    # Désérialiser la chaîne pour obtenir une vraie liste d'objets Python
    customers_data = json.loads(raw_data)

# Afficher le premier client (index 1)
customers_data[1]
value = customers_data[0]
key = value['customer_id']
print(key,value)
```

```python
type(str(value).encode('utf-8'))
```

```python
# send that to Kafka
producer.produce(topic, key=str(key).encode('utf-8'), value=str(value).encode('utf-8'))
```

```python
# Send multiple messages to our Kafka Cluster
```

```python
def delivery_status(err, msg):
    if err:
        print('Error: Message failed delivery: {}'.format(err))
    else:
        print('Message delivered to {} [{}]'.format(msg.topic(), msg.partition()))

for record in customers_data:
    try:
        message_value = json.dumps(record)
        # Correction : Convertir l'ID entier en chaîne avant de l'encoder
        message_key = str(record['customer_id']).encode('utf-8')

        producer.produce(topic, key=message_key, value=message_value, callback=delivery_status)
        producer.poll(0.1)

    except Exception as e:
        print(f"Error sending messages: {e}")

producer.flush()
print("All messages sent to Kafka successfully")
```
